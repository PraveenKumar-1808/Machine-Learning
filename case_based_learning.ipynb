```python
# Previous cases
cases = [
    {"hours": 2, "mark": 45},
    {"hours": 4, "mark": 60},
    {"hours": 6, "mark": 75},
    {"hours": 8, "mark": 88}
]

# New student
new_hours = 5.5

# Find the most similar previous case
best_case = None
smallest_distance = float("inf")

for case in cases:
    distance = abs(case["hours"] - new_hours)

    print(
        "Previous Case:",
        case["hours"],
        "hours | Distance:",
        distance
    )

    if distance < smallest_distance:
        smallest_distance = distance
        best_case = case

print("\nNew Student:", new_hours, "hours")

print(
    "Most Similar Previous Case:",
    best_case["hours"],
    "hours"
)

print(
    "Previous Student Mark:",
    best_case["mark"]
)

print(
    "Suggested Predicted Mark:",
    best_case["mark"]
)
```
