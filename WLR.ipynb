```python
import numpy as np
import matplotlib.pyplot as plt

# Placeholder definitions for missing variables
# In a full implementation, these would come from your dataset and specific query.
X = np.array([1, 2, 3, 4, 5]) # Example feature values
y = np.array([2, 4, 5, 4, 5]) # Example target values
query = 3.5 # Example query point

# For simplicity, X_design and weights are also placeholders.
# In a real LWR, X_design is often [1, X] and weights are calculated based on query and X.
X_design = np.vstack([np.ones(len(X)), X]).T
weights = np.ones(len(X)) # Example: all weights are 1. In real LWR, these would vary.

W = np.diag(weights)

theta = np.linalg.pinv(
    X_design.T @ W @ X_design
    ) @ X_design.T @ W @ y

# Prediction
prediction = np.array([1, query]) @ theta

print("\nNew student studies:", query, "hours")
print("Predicted Mark:", round(prediction, 2))

# Plot
plt.scatter(X, y, label="Students")
plt.scatter(query, prediction, s=120, label="New Student")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Locally Weighted Regression")
plt.legend()
plt.show()
```
